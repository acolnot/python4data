# 1. Importing Data From A `.csv` File

```python
# 1. Define column names
columns = ['animal_type', 'intake_year', 'intake_condition', 'intake_number', 'intake_type', 'sex_upon_intake', \
          'age_upon_intake_(years)', 'time_in_shelter_days', 'sex_upon_outcome', 'age_upon_outcome_(years)', \
          'outcome_type']

# 2. Load dataset
data = pd.read_csv('./data/aac_intakes_outcomes.csv', usecols=columns)

# 3. Visualize data /!\ .head() must be the last line of the cell, otherwise  it is 
# --> to force visualization: use display()
data.head()
```

# 3. Operations on the Dataset

## 3.1 Drop (= remove) rows/columns

By default it drops rows (axis = 0).

```python
data_no_intake_type = data.drop('intake_type', axis=1)
data_no_intake_type.head()
```

## 3.2 Apply Lambdas

```python
# Create a new column in the dataset by applying a lambda
data['age_in_months'] = data['age_upon_intake_(years)'].apply(lambda x: x * 12)
data.head()
```

## 3.3. Conditional Filtering & Slicing

```python
# Filter the dataset to look only at rows where the animal type is 'Dog'
dogs_data = data.loc[data['animal_type'] == 'Dog']
dogs_data.head()
```

## 3.4. Frequencies `value_counts()`

```python
# See how many animals of each type are in the dataset
data['animal_type'].value_counts()
```

## 3.5. Split Data `groupby()`

```python
# Group the dataset by animal type
grouped_by_animal = data.groupby('animal_type')
display(grouped_by_animal)

# Count how many records exist for each animal type group
animal_counts = grouped_by_animal.size()

display(animal_counts)
```

## 3.6. Handle Missing Values

### 3.6.1. `fillna()`

Replaces missing (NaN or null) values in the dataset with a specified scalar value, a dictionary, or a computed metric (like a column's mean or median).

```python
# Fill missing age values with the average (mean) age of the column
mean_age = data['age_upon_intake_(years)'].mean()
data['age_upon_intake_filled'] = data['age_upon_intake_(years)'].fillna(mean_age)
data[['age_upon_intake_filled']].head()
```
