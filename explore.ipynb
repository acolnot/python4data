## 3.9. Dataset Restructuration

### 3.9.1. `.merge()`

Performs a relational join  between two DataFrames based on matching values in specified columns or indices. 

Mathematically, it calculates the Cartesian product of matching keys and filters them based on the specified join type:

- `inner`: Keeps only the rows that have a matching key in both tables. If a value exists in Table A but has no counterpart in Table B, it is completely dropped from the final output.

- `left`: Keeps 100% of the rows from the Left table. It pulls in matching data from Table B where available; where it isn't available, it inserts NaN.

- `right`: Keeps 100% of the rows from the Right table (Table B). It pulls in matching data from Table A where available; where it isn't available, it inserts NaN.

- `outer`: Keeps all rows from both tables combined. If a row exists in one table but not the other, it is still included, and NaN values fill in the missing gaps for the missing side.

```python
# Create a new df combining columns of both data & extra_info
extra_info = pd.DataFrame({'intake_type': ['Stray', 'Public Assist', 'Owner Surrender'], 
                           'priority_level': ['High', 'Medium', 'Low']})

# Merge the auxiliary info into your main dataset based on 'intake_type'
merged_data = pd.merge(data, extra_info, on='intake_type', how='left')
merged_data.head()
```

### 3.9.2. `.stack()`

Pivots or rotates the columns of a DataFrame into the innermost level of the row index, transforming a wide-format layout into a long-format hierarchical Series. 

Mathematically, it reshapes the column axis index labels into a multi-index row layout.

To use when we need to compress multi-column metrics down into a single indexed column for multi-level data manipulation.

```python
# Create a small DataFrame 
summary_df = data.groupby('animal_type')[['age_upon_intake_(years)', 'time_in_shelter_days']].mean()

print("--- Original DataFrame ---")
display(summary_df)

# Use .stack() to pivot the column headers into an extra row index layer
stacked_series = summary_df.stack()

print("\n--- After .stack() ---")
display(stacked_series)
```

### 3.9.3. `.pivot()`

Reshapes data based on column values, turning unique values from a single column into entirely new distinct columns (wide format). 

Mathematically, it maps a long-form relation table into a wide coordinate matrix.

To use when we want to transform long-format rows into a clear, wide cross-tabulated matrix for easier visual comparisons.

```python
# Pivot a small slice of data to show intake years as columns for each index position
sample_df = data.head(10).reset_index()
sample_df.pivot(index='index', columns='animal_type', values='intake_year')
```

### 3.9.4 `.melt()`

Unpivots a DataFrame from a wide format into a long format, consolidating multiple attribute columns into two unified columns: variable and value. 

Mathematically, it performs the inverse operation of a pivot.

To use when multiple columns represent the same type of variable and you need them stacked into a single column for analysis or visualization.

```python
# Melt intake and outcome age columns into a single long-format attribute list
pd.melt(data.head(5), id_vars=['animal_type'], value_vars=['age_upon_intake_(years)', 'age_upon_outcome_(years)'])
```
